# Solving linear systems in practice

Linear systems of the form:

$$
\begin{split}
\Amat \xvec = \yvec
\end{split}
$$

are never solved by explicitly inverting $\Amat$. What we use is LU factorization of $\Amat$ and then solve using forward and backward substitution. This is what ```numpy.linalg.solve``` does. Other libraries, such as PyTorch, have different routines to solve these systems when $\Amat$ has special structure. For instance, if $\Amat$ is positive definite, then there are routines to solve this directly from the Cholesky factorization. 

For our purpose, for instance, the solution of the ordinary least squares method  is never obtained by inverting $\Xmatt \Xmat$. Rather, we use ```np.linalg.solve``` as follows:

```python
np.linalg.solve(np.transpose(X)@X,np.transpose(X)@t) 
```

Under the hood, NumPy (through LAPACK and BLAS) performs the LU factorization and applies forward and backward substitution. Other software might take advantage if one directly has the Cholesky factorization (when possible) of any of the matrices. PyTorch, for instance, supports many ways of solving these systems. NumPy, as far as I know, has only this one.
