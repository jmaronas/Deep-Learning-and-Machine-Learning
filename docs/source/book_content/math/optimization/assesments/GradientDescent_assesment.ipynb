````python
%matplotlib inline

# Create temporary file for video creation
video_filename = "/tmp/aux.mp4"

## video writer
writer = imageio.get_writer(video_filename, format="FFMPEG", mode="I", fps=1, codec="libx264")

## ===================
## plot configurations
N_grid = 100

min_x = -6
max_x = 6
x_grid = np.linspace(min_x,max_x,N_grid)

## ======
## figure
fig, (ax1, ax2) = plt.subplots(1,2, figsize = (10,5))

    
## ========
## initialization point and grad descent parameters
x_t = -5
lr = 0.1
epochs = 10

# to keep history of updates
history_fx = []
history_epochs = []

for e in range(epochs):
    
    ## compute gradient
    grad_fx = grad_fun(x_t)
    
    ## compute function 
    f_x = fun(x_t)

    ## grad descent update
    x_n = x_t - lr * grad_fx
    
    ## =============
    ## Plot function minimization
    ax1.clear()
    ax1.set_xlim([-7,7])
    ax1.set_ylim([-5,40])
    
    ## 1. display function and current point
    ax1.plot(x_grid, fun(x_grid), label = 'function')
    ax1.plot(x_t, -1,'*', color = 'C1')
    ax1.legend(loc = "upper right")

    ## 2. display current function
    ax1.plot(x_t, f_x, 'o', color = 'C1', label = f'$f(x_t)$', zorder = 20)
    ax1.text(x_t + 0.5, f_x , f"$f(x_t)={f_x:.3f}$", fontsize=12, va='bottom', color = "C0" , zorder = 50)
    ax1.text(x_t + 0.5, -4, f"$x_{{t}}$ = {x_t:.2f}", va='bottom', color = "C1" , zorder = 200) 
    ax1.legend(loc = "upper right")

    ## 3. display new point
    ax1.plot(x_n, -1, '*', color = 'C3', label = r'updated value: $x_t = x - \alpha\cdot\frac{\partial f(x)}{\partial x}$', zorder = 200, markersize = 10)
    ax1.text(x_n+0.5, -1, f"$x_{{t+1}} = {x_t:.2f} -{lr:.2f}*{grad_fx:.2f} = {x_n:.2f}$", va='bottom', color = f"C3" , zorder = 200) 
    ax1.legend(loc = "upper right")
    
    ax1.set_title(f"$f(x_t)={f_x:.3f}$ \n $f(x_{{t+1}}) = {fun(x_n):.3f}$")
    ax1.set_xlabel('$x$')
    ax1.set_ylabel('$f(x)$')
    
    # to repeat loop
    x_t = x_n
 
    ## =============
    ## Plot function minimization over the course of learning
    ax2.clear()
    history_epochs.append(e)
    history_fx.append(f_x)
    
    ax2.plot(history_epochs, history_fx,'o', color = 'k')
    ax2.plot(history_epochs, history_fx, color = 'k')
    
    ax2.set_xlabel('Epochs')
    ax2.set_ylabel('$f(x)$')
    ax2.set_xlim([-1,epochs+1])
    ax2.set_ylim([0,30])

    fig.canvas.draw()
    frame = np.asarray(fig.canvas.buffer_rgba())[:, :, :3]
    writer.append_data(frame)

writer.close() 
plt.close()
````

## Multidimensional functions

Consider the function:

$$
\begin{split}
f(x,y) = x^4 - 2x^2 + 2y^2
\end{split}
$$

**Task 4:**

As before, implement two Python functions that implement the function and the gradient. The gradient function must return the gradient over each coordinate.

```python
def fun( x:np.array, y:np.array ):
    ...

def grad_fun(x,y):
    return grad_x, grad_y
```

**Solution:**

````python
def fun(x,y):
    return x**4 -2*x**2 + 2*y**2

def grad_fun(x,y):
    return 4*x**3 - 4*x, 4*y
````

**Solution**

````python
## define grid
N_grid = 100
min_x = -2
max_x = 2
min_y = -2
max_y = 2

x_grid = np.linspace(min_x,max_x,N_grid)
y_grid = np.linspace(min_y, max_y,N_grid)

X_mesh, Y_mesh = np.meshgrid(x_grid, y_grid)
Z_mesh = fun(X_mesh,Y_mesh)

## define plot
fig = plt.figure(figsize=(10, 5))
ax1 = fig.add_subplot(1, 2, 1)
ax2 = fig.add_subplot(1, 2, 2, projection='3d')
ax2.view_init(elev=65)

ax1.contourf(X_mesh, Y_mesh, Z_mesh, cmap = 'gray')
ax1.set_xlabel('$x$')
ax1.set_ylabel('$y$')

ax2.plot_surface(X_mesh, Y_mesh, Z_mesh, alpha=0.8, cmap = 'gray')
ax2.set_xlabel(r'$x$')
ax2.set_ylabel(r'$y$')
ax2.set_zlabel(r'$f(x,y)$')
````

**Solution:**

````python
## define grid
N_grid = 100
min_x = -2
max_x = 2
min_y = -2
max_y = 2

x_grid = np.linspace(min_x,max_x,N_grid)
y_grid = np.linspace(min_y, max_y,N_grid)

X_mesh, Y_mesh = np.meshgrid(x_grid, y_grid)
Z_mesh = fun(X_mesh,Y_mesh)

## define plot
fig = plt.figure(figsize=(10, 5))
ax1 = fig.add_subplot(1, 2, 1)
ax2 = fig.add_subplot(1, 2, 2)

# Initialize parameters
x = -1.7
y = -1.7

## gradient descent parameters
lr = 0.1
epochs = 10

# Create a temporary file for video creation
video_filename = "/tmp/aux.mp4"

## video writer
writer = imageio.get_writer(video_filename, format="FFMPEG", mode="I", fps=1, codec="libx264")

history_fx = []
history_epochs = []

for e in range(epochs):

    ## forward plus backward
    grad_x, grad_y = grad_fun(x,y)

    ## compute function at current parameter value
    f_xy = fun(x,y)

    ## compute function on updated parameters
    x_n = x-lr*grad_x
    y_n = y-lr*grad_y

    ## function on new parameters
    f_xy_n = fun(x_n,y_n)

    ## ============= ##
    ## START DRAWING ##
    ## ============= ##
    ax1.clear()

    ## 1 display function
    ax1.contourf(X_mesh,Y_mesh,Z_mesh, cmap = 'gray')
    
    ## 2 display current point
    ax1.plot(x,y, '*', color='C1')
    ax1.text(x, y, f"$(x_t,y_t)$ = ({x:.2f},{y:.2f})", fontsize=12, va='bottom', color="C1")

    ## save frame
    fig.canvas.draw()
    frame = np.asarray(fig.canvas.buffer_rgba())[:, :, :3]
    writer.append_data(frame)
    
    ## 3 display new weight
    ax1.plot(x_n, y_n, 'o', color='C3')
    ax1.text(x_n, y_n, f"$(x_{{t+1}},y_{{t+1}}) = ({x_n:.2f},{y_n:.2f})$", fontsize=12, va='bottom', color="C3")

    ## save frame
    fig.canvas.draw()
    frame = np.asarray(fig.canvas.buffer_rgba())[:, :, :3]
    writer.append_data(frame)
    
    history_epochs.append(e)
    history_fx.append(f_xy_n)
    
    ax2.plot(history_epochs, history_fx,'o', color = 'k')
    ax2.plot(history_epochs, history_fx, color = 'k')
    
    ax2.set_xlabel('Epochs')
    ax2.set_ylabel('$f(x)$')
    ax2.set_xlim([-1,epochs+1])
    ax2.set_ylim([-1.5,2.5])
    
    ## save frame
    fig.canvas.draw()
    frame = np.asarray(fig.canvas.buffer_rgba())[:, :, :3]
    writer.append_data(frame)

    x = x_n
    y = y_n

writer.close()
plt.close()
    
````
