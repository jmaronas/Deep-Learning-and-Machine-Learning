# Critical points of functions

We know from high school that obtaining the minima or maxima (known as critical points) of a function can be done by solving the system of equations that results from finding the variable values that make the gradient 0. A function can have one or more than one of these points. We also have saddle points, which are points where the gradient is also 0, but they are not minima nor maxima.

Determine if a point is a minimum, a maximum, or a saddle point is done through the second derivative information, contained in the Hessian for multivariate functions and in the standard second derivative for univariate functions.

In other words, given $\phi: \mathbb{R}^N \to \mathbb{R}$ (when the image is a vector, the concept of minimum is not defined in the same way, at least with my knowledge), a critical point is given by solving:

$$
\begin{split}
\grad{\xvec}{} \phi(\xvec) = 0\\
\end{split}
$$

In functions $f:\mathbb{R}\to \mathbb{R}$, the second derivative tells us whether it is a minimum or a maximum. There are no saddle points in these cases, and they are named inflection points (points where curvature changes). When the second derivative is positive, we have a minimum; when it is negative, a maximum; and when it is 0, we need further analysis. 

In a multivariate function, we need to obtain the Hessian. When the Hessian is positive definite (all eigenvalues are positive at the point), we have a local minimum. When the Hessian is negative definite (all eigenvalues are negative), we have a maximum. When it has both positive and negative eigenvalues, then it is a saddle point. 

When we have $\phi:\mathbb{R}^2 \to \mathbb{R}$, then the determinant can be used to obtain this information.

Importantly, finding the minimum of a function $\phi(x)$ is equivalent to finding the maximum of the function $-\phi(x)$.

## Types of functions

### Scalar-Valued Functions

A scalar-valued function is a function that takes as input any argument (scalar, matrix, or vector) and returns a scalar value. Examples of such functions are given by:

$$
\begin{split}
\phi(x) = (x^2 - 4)^3 + 2\\
\phi(\xvec) = \xvect \Amat \xvec\\
\phi(\Xmat) = \tr{\Xmatinv\Amat}
\end{split}
$$

These are commonly denoted as:

$$
\begin{split}
\phi : \mathbb{R} \to \mathbb{R}\\
\phi : \mathbb{R}^n \to \mathbb{R}\\
\phi : \mathbb{R}^{n\times n} \to \mathbb{R}
\end{split}
$$

Let's visualize some examples

#### $\phi:\mathbb{R} \to \mathbb{R}$

```python
N_grid = 100
min_x = -2
max_x = 2
x_grid = np.linspace(min_x,max_x,N_grid)

fig, ((ax1,ax2),(ax3,ax4)) = plt.subplots(2,2, figsize = (10,10))

ax1.plot(x_grid, x_grid**2)
ax1.set_title(f"$x^2$")

ax2.plot(x_grid, x_grid**3)
ax2.set_title(f"$x^3$")

ax3.plot(x_grid, x_grid**4)
ax3.set_title(f"$x^4$")

ax4.plot(x_grid, x_grid**4 - 2*x_grid**2 )
ax4.set_title(f"$x^4 - 2 x^2$")
```

### Vector-Valued Functions

A vector-valued function is a function that takes as input any argument (scalar, matrix, or vector) and returns a vector. Examples of such functions are given by:

$$
\begin{split}
\phivec(x) = [\cos(x),\sin(x)]\\
\phivec(\xvec) = \avec[\xvect \Amat \xvec]\\
\phivec(\Xmat) = \avec\Xmatinv\Amat
\end{split}
$$

These are commonly denoted as:

$$
\begin{split}
\phi : \mathbb{R} \to \mathbb{R}^m\\
\phi : \mathbb{R}^n \to \mathbb{R}^m\\
\phi : \mathbb{R}^{n\times n'} \to \mathbb{R}^m
\end{split}
$$

The only one we can visualize is $\phi : \mathbb{R} \to \mathbb{R}^2$

```python
def f(t):
    x = t*np.cos(t)
    y = t*np.sin(t)
    return x, y

# parameter
t = np.linspace(0, 2*np.pi, 500)

# function
x, y = f(t)

# 2D and 3D plot
fig = plt.figure(figsize=(12, 5))

ax1 = fig.add_subplot(1, 2, 1)
ax1.plot(x, y)
ax1.set_title(r"$t\cos(t),t\sin(t)$")
ax1.set_xlabel("x(t)")
ax1.set_ylabel("y(t)")
ax1.axhline(0, color='black', linewidth=0.5)
ax1.axvline(0, color='black', linewidth=0.5)
ax1.set_aspect('equal')
ax1.grid(True)

# 3D plot
ax2 = fig.add_subplot(1, 2, 2, projection='3d')

ax2.plot(x, y, t)
ax2.set_title(r"$t\cos(t),t\sin(t)$")
ax2.set_xlabel("x(t)")
ax2.set_ylabel("y(t)")
ax2.set_zlabel("t")
```

### Special functions

#### Sumations

Summations are a special case of scalar-value functions. Something important to note is that a summation of $N$ elements is a vector-argument scalar-value function. Suppose we have a vector of $N$ elements, then if we define as $\onevec$ a vector containing ones, then:

$$
\begin{split}
\onevect \xvec = \sum^N_{n=1} x_n
\end{split}
$$

This is important since it allows us to express in a compact and clean way a function that, in principle, can have an arbitrary number of arguments (depending on how many elements we might need to sum). 

Note that we might have vector or matrix argument functions that might not be written down compactly using standard algebra operators. For example, the product of the elements in $\xvec$ or, simply:

$$
\begin{split}
\phi(\xvec) = \sin(x_1) + x_2^2
\end{split}
$$

#### Element-wise functions

An element-wise function is a function that operates on structured inputs (such as vectors, matrices, or tensors) by applying the same scalar function independently to each individual entry, returning an output of the same shape.

$$
\phivec(\xvec) = \xvec^2, \quad
\xvec =
\begin{pmatrix}
1 \\
2 \\
3
\end{pmatrix}
\Rightarrow
\phivec(\xvec) =
\begin{pmatrix}
1^2 \\
2^2 \\
3^2
\end{pmatrix}
=
\begin{pmatrix}
1 \\
4 \\
9
\end{pmatrix}
$$

$$
\phimat(\Xmat) = \Xmat^2, \quad
\Xmat =
\begin{pmatrix}
1 & 2 & 3 \\
4 & 5 & 6 \\
7 & 8 & 9
\end{pmatrix}
\Rightarrow
\phimat(\Xmat) =
\begin{pmatrix}
1^2 & 2^2 & 3^2 \\
4^2 & 5^2 & 6^2 \\
7^2 & 8^2 & 9^2
\end{pmatrix}
=
\begin{pmatrix}
1 & 4 & 9 \\
16 & 25 & 36 \\
49 & 64 & 81
\end{pmatrix}
$$

Common scalar functions applied are the logarithm, exponential, square root, etc. In reality, any scalar-valued function of a scalar argument can be used as an element-wise function.

## Gradient

For scalar-valued functions, the gradient is a column vector of the partial derivatives of the function wrt the arguments.

$$
\begin{split}
\grad{x}{} \phi(x) = \frac{\dd \phi(x)}{\dd x} 
\end{split}
$$

$$
\begin{split}
\grad{\xvec}{} \phi(\xvec) = \begin{pmatrix}
\frac{\partial \phi}{\partial x_1} \\
\frac{\partial \phi}{\partial x_2} \\
\vdots\\
\frac{\partial \phi}{\partial x_n}
\end{pmatrix}
\end{split}
$$

The matrix argument is a special case I am not covering, yet it is based on representing a matrix as a vector that results from stacking its columns and then using standard calculus. Anyone interested can start learning from here https://arxiv.org/pdf/2506.23996.

The gradient is a vector field expressing the direction of change of a function in all its directions. Let's visualize the gradient over our previous scalar value functions.

```python
N_grid = 100
min_x = -2
max_x = 2
x_grid = np.linspace(min_x, max_x, N_grid)

fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(10, 10))

# --- f(x) = x^2 ---
ax1.plot(x_grid, x_grid**2, label=r"$x^2$")
ax1.plot(x_grid, 2*x_grid, label=r"$\nabla x^2 = 2x$")
ax1.set_title(r"$x^2$")
ax1.legend()

# --- f(x) = x^3 ---
ax2.plot(x_grid, x_grid**3, label=r"$x^3$")
ax2.plot(x_grid, 3*x_grid**2, label=r"$\nabla x^3 = 3x^2$")
ax2.set_title(r"$x^3$")
ax2.legend()

# --- f(x) = x^4 ---
ax3.plot(x_grid, x_grid**4, label=r"$x^4$")
ax3.plot(x_grid, 4*x_grid**3, label=r"$\nabla x^4 = 4x^3$")
ax3.set_title(r"$x^4$")
ax3.legend()

# --- f(x) = x^4 - 2x^2 ---
ax4.plot(x_grid, x_grid**4 - 2*x_grid**2, label=r"$x^4 - 2x^2$")
ax4.plot(x_grid, 4*x_grid**3 - 4*x_grid, label=r"$\nabla (x^4 - 2x^2) = 4x^3 - 4x$")
ax4.set_title(r"$x^4 - 2x^2$")
ax4.legend()

plt.tight_layout()
plt.show()
```

For each of these cases, we have:

### $x^2$

- derivative: $2x$

- critical point: $2x = 0 \to x = 0$

- second derivative: $2>0 \to $ minimum 

### $x^3$

- derivative: $3x^2$

- critical point: $3x^2 = 0 \to x = 0$

- second derivative: $6x$. At $x=0$, the value is $0$, hence identifying as an inflection point, a change of curvature. It is negative for values $x<0$ and positive for the rest, which implies a change of curvature at that point but not a minimum or maximum. 

### $x^4$

- derivative: $4x^3$

- critical point: $4x^3 \to x = 0$

- second derivative: $12x^2 > 0 \to $ minimum.

### $x^4 - 2x^2$

- derivative: $4x^3-4x$

- critical point: $4x^3-4x = 0 \to x = 0, x = 1, x = -1$

- second derivative: $12x^2 -4$. Giving values, we have: 

$$
\begin{split}
12x^2 -4 < 0, x = 0 \to \text{maximum}\\
12x^2 -4 > 0, x = -1 \to \text{minimum}\\
12x^2 -4 > 0, x = 1 \to \text{minimum}
\end{split}
$$



For multidimensional functions, we can express the gradient at each point as a vector that points in the steepest direction. Clearly, the gradient represents a vector field.

```python
def fun1(x,y):
    return x**2 - y**2

def fun2(x,y):
    return x**2 + y**2

def fun3(x,y):
    return -x**2 - y**2

def fun4(x,y):
    return x**4 -2*x**2 + 2*y**2


## define gradients
def grad_fun1(x,y):
    return 2*x, -2*y

def grad_fun2(x,y):
    return 2*x, 2*y

def grad_fun3(x,y):
    return -2*x, -2*y

def grad_fun4(x,y):
    return 4*x**3 - 4*x, 4*y


## define grid
N_grid = 25
min_x = -2
max_x = 2
min_y = -2
max_y = 2

x_grid = np.linspace(min_x,max_x,N_grid)
y_grid = np.linspace(min_y, max_y, N_grid)

X_mesh, Y_mesh = np.meshgrid(x_grid, y_grid)

## define plot
fig = plt.figure(figsize=(12, 16))
ax_list = []

for i in range(4):
    ax1 = fig.add_subplot(4, 2, 2*i+1)
    ax2 = fig.add_subplot(4, 2, 2*i+2, projection='3d')
    
    ax_list.append([ax1,ax2])

# define set of func to show
funcs = [fun1, fun2, fun3, fun4]
grads = [grad_fun1, grad_fun2, grad_fun3, grad_fun4]

titles = [
    r"$x^2 - y^2$",
    r"$x^2 + y^2$",
    r"$-x^2 - y^2$",
    r"$x^4 - 2x^2 + 2y^2$"
]

for i in range(4):
    ax1, ax2 = ax_list[i]
    
    Z_mesh = funcs[i](X_mesh,Y_mesh)
    U_mesh, V_mesh = grads[i](X_mesh,Y_mesh)
    
    ## contour plot + gradient field
    ax1.contourf(X_mesh,Y_mesh,Z_mesh, levels=50, cmap='gray')
    ax1.quiver(X_mesh,Y_mesh,U_mesh,V_mesh,color='red',alpha=0.7)

    ## surface plot
    ax2.plot_surface(X_mesh,Y_mesh,Z_mesh,cmap='gray')
    
    ## plot tangent vectors over 3d plots
    step = 2

    Xq = X_mesh[::step, ::step]
    Yq = Y_mesh[::step, ::step]
    Zq = Z_mesh[::step, ::step]
    
    Uq = U_mesh[::step, ::step]
    Vq = V_mesh[::step, ::step]
    

    Wq = Uq**2 + Vq**2
    
    ax2.quiver(
        Xq, Yq, Zq,   
        Uq, Vq, Wq,  
        color='red',
        normalize=True
    )

    # titles and so on
    ax1.set_title(titles[i])
    
    ax2.set_title(titles[i])
```

Similar to the one-dimensional case, the critical points are obtained by solving the system of equations that results from taking the gradient and equating it to 0.

## Jacobian

The gradient of a vector-valued function is not defined. Here, we talk about the Jacobian. We care about the Jacobian for two reasons:

1) As mentioned is the generalization of the gradient to any type of function
2) It is the element that is used in the chain rule

The Jacobian matrix is a matrix of partial derivatives. Since matrix-argument functions are really cast into vector-argument functions, we just need the definition of the vector-valued case. For the general case of vector-valued functions $\phi:\mathbb{R}^n\to \mathbb{R}^m$, the Jacobian matrix is given by:

$$
\Jac{\xvec}{}\phivec(\xvec) =
\begin{pmatrix}
\frac{\partial \phi_1}{\partial x_1} & \frac{\partial \phi_1}{\partial x_2} & \cdots & \frac{\partial \phi_1}{\partial x_n} \\
\frac{\partial \phi_2}{\partial x_1} & \frac{\partial \phi_2}{\partial x_2} & \cdots & \frac{\partial \phi_2}{\partial x_n} \\
\vdots & \vdots & \ddots & \vdots \\
\frac{\partial \phi_m}{\partial x_1} & \frac{\partial \phi_m}{\partial x_2} & \cdots & \frac{\partial \phi_m}{\partial x_n}
\end{pmatrix}
$$

### Special cases: 

* When $m=1$ and $n\neq 1$, the Jacobian is a row vector, which is exactly the transpose of the gradient, which is a column vector:

$$
\Jac{\xvec}{\phi}  =
\begin{pmatrix}
\frac{\partial \phi}{\partial x_1} & \frac{\partial \phi}{\partial x_2} & \cdots & \frac{\partial \phi}{\partial x_n}
\end{pmatrix}
$$

Thus, moving from the Jacobian to the gradient is simply done by transposition. 

* When $m=1$ and $n=1$, the Jacobian is a scalar value, known as the derivative:

$$
\Jac{x}{\phi} =
\frac{\dd \phi}{\dd x} 
$$ 

* When $n=1$ and $m\neq1$, the Jacobian is a column vector. Importantly, this is the case of a vector-valued function of a scalar argument function so this is not the gradient. The gradient is only defined for scalar-valued functions.


## Chain Rule

The question now is: how do we obtain the derivative of whatever function?. Suppose that we want to obtain the derivative of these functions, wrt $\xvec$:

$$
\begin{split}
\phi(x) &= (x^2 - 4)^3 + 2\\
\phi(\xvec) &= \onevect\log(\Wmat\xvec);\quad  \Wmat \in \mathbb{R}^{q \times n},\xvec \in \mathbb{R}^n
\end{split}
$$

The first one looks simple to obtain, using standard high-school methods. The second one would require us to expand the linear combination, apply the logarithm to each of the elements in the resulting $ n$-dimensional vector, and compute the sum. Once the sum is expressed as a function of each individual element in $\xvec$, compute all the partial derivatives. Finally, by inspection, write down the general form. This does not look like something practical.  While the first case looks simple, there might be other scalar argument functions that do not look so easy to differentiate, for instance:

$$
\begin{split}
\phi(x) = \frac{x}{1+e^{-\log x}}
\end{split}
$$

The chain rule is a mathematical tool that allows us to obtain the derivative of a function wrt any variable using a systematic procedure. 

In reality, the chain rule is a tool to obtain the Jacobian of a composition of functions. So when we have a complex function, we can just break it down and express it through a composition of simple functions from which the Jacobian is easily computed.

Attending to the chain rule, the Jacobian of a function can be obtained by multiplying the Jacobians of the individual transformations. Let's see some examples:

### Examples

#### Case $m=1$, $n=1$

Consider the function:

$$
y = (x^2 - 4)^3 + 2
$$

This function can be written as:

$$
\begin{split}
t = x^2 \\
u = t - 4\\
z = u^3 \\
y = z + 2 
\end{split}
$$

Let's start substituting to show it is equivalent. First subsitute $z = u^3$ into $y = z + 2$, yielding $y = u^3 + 2$. Now do the same thing and substitute $u = t - 4$ into $y = u^3 + 2$, yielding: $y = (t - 4)^3 + 2$, and finally subsitute $t = x^2$ into $y = (t - 4)^3 + 2$ yielding $y = ( x^2 - 4)^3 + 2$, which is our original function.

When broken into small functions, we know that the full derivative can be obtained by multiplying the derivatives of the individual functions. In other words:

$$
\frac{\dd y}{\dd x} = \frac{\dd y}{\dd z}\frac{\dd z}{\dd u}\frac{\dd u}{\dd t}\frac{\dd t}{\dd x}
$$

$$
\begin{split}
&\frac{\dd y}{\dd z} = 1\\
&\frac{\dd z}{\dd u} = 3u^2\\
&\frac{\dd u}{\dd t} = 1\\
&\frac{\dd t}{\dd x} = 2x\\
\end{split}
$$

So, applying the chain rule, we have:

$$
\frac{\dd y}{\dd x} = \frac{\dd y}{\dd z}\frac{\dd z}{\dd u}\frac{\dd u}{\dd t}\frac{\dd t}{\dd x} = 1 \cdot 3u^2 \cdot 1 \cdot 2x = 3u^2\cdot 2x 
$$

Finally substituting the value of $u=x^2-4$ depending on $x$ we have:

$$
\frac{\dd y}{\dd x} = 3(x^2-4)^2 2x
$$

You can check that this is exactly the value you obtain when you differentiate the function automatically, as you were taught in high school.

#### Case $m=1$, $n\neq 1$

Suppose that we now deal with:

$$
\begin{split}
\phi(\xvec) = \onevect\exp(\Wmat\xvec); \quad \Wmat \in \mathbb{R}^{q \times n},\xvec \in \mathbb{R}^n
\end{split}
$$

where $\exp$ denotes an element-wise exponential function. We can express this function through the following composition.

$$
\begin{split}
\zvec &=  \Wmat\xvec; \quad \zvec \in \mathbb{R}^q  \quad \phi : \mathbb{R}^n \to \mathbb{R}^q \\
\avec &= \exp(\zvec); \quad \avec \in \mathbb{R}^q  \quad \phi : \mathbb{R}^q \to \mathbb{R}^q \\
\phi(\xvec) &= \onevect\avec;  \quad\phi(\xvec) \in \mathbb{R} \quad \phi : \mathbb{R}^q \to \mathbb{R}
\end{split}
$$

Given that we know the input and output dimensionality of each of the individual functions of the composition, we know the shape of the Jacobians. The first function, for example, takes as input a vector and outputs a vector; thus the Jacobian is an $q\times n$ matrix. The Jacobian of the second function is a $q\times q$ matrix. Finally, the last composition is a scalar function, so the Jacobian is a row vector. 

The Jacobians of each of these transformations are:

$$
\begin{split}
\Jac{\xvec}{\zvec} &= \Wmat \in \mathbb{R}^{q\times n}\\
\Jac{\zvec}{\avec} &= \diag (\exp(\zvec)); \in \mathbb{R}^{q\times q} \\
\Jac{\avec}{\phi} &= \onevect  \in \mathbb{R}^{1 \times q}
\end{split}
$$

Thus, the Jacobian of the transformation is given by the product:

$$
\begin{split}
\Jac{\xvec}\phi(\xvec) &=  \onevect\diag (\exp(\zvec))\Wmat\\
&=  \onevect\diag (\exp(\Wmat\xvec))\Wmat \in \mathbb{R}^{1 \times n}
\end{split}
$$

Note that this is a Jacobian, which is an n-dimensional row vector, as expected.

To see that the above derivation is true, consider, for simplicity, $m=2$ and $n=2$; then we have:

$$
\begin{split}
\Wmat=\begin{pmatrix}
w_{11} & w_{12} \\
w_{21} & w_{22}
\end{pmatrix}\\
\xvect = [x_1,x_2]
\end{split}
$$

Then we can express the previous function as:

$$
\begin{split}
\phi(\xvec) =
\exp(w_{11}x_1 + w_{12}x_2)
+
\exp(w_{21}x_1 + w_{22}x_2)
\end{split}
$$

Now, if you want, take each of the partial derivatives wrt $x_1$ and $x_2$, and you will see that that is exactly what gives my derivation: $\onevect\exp(\Wmat\xvec)\Wmat$. Obviously, the general derivative case implies that $n$ and $m$ take arbitrary values. Once you have done the previous exercise, try to do the same for:

$$
\begin{split}
\phi(\mathbf{\xvec}) = \sum_{i=1}^{m} \exp\!\left(\sum_{j=1}^{n} w_{ij} x_j \right)
\end{split}
$$

You will see that the expression of this derivative is exactly what I wrote.  Now, you believe in the power of the chain rule. Lucky for you, or not, machine learning is all about this kind of function composition. 

#### Case $m=1$, $n=1$ through arbitrary shape

One curious thing about the kind of function compositions that arise in machine learning is that we sometimes deal with functions that take as input a single scalar and return a single scalar, but that, through the composition, are made up from vector-valued functions. This is the typical case of many regression models.

Consider now the function $\phi:\mathbb{R}\to\mathbb{R}$:

$$
\begin{split}
\phi(x) = \onevect \frac{1}{1+e^{- \wvec x}}; \wvec \in \mathbb{R}^{q}, x \in \mathbb{R}
\end{split}
$$

Again, the exponential function here acts as an elementwise function. This looks like a very ugly function. In turn, the expansion of this function is given by:

$$
\begin{split}
\phi(x) =
\sum_{i=1}^{q}
\frac{1}{1 + e^{-w_i x}}
\end{split}
$$

So the derivative wrt $x$ is the sum of the derivatives of $\frac{1}{1 + e^{-w_i x}}$. With our use of the chain rule, we can break this down as the following composition:

$$
\begin{split}
\zvec = \wvec x \\
\yvec = \frac{1}{1+e^{-\zvec}} \\
\phi(x) =  \onevect \yvec
\end{split}
$$

The Jacobians of each of the transformations are given by:

$$
\begin{split}
\Jac{x}{\zvec} &= \wvec\\
\Jac{\zvec}{\yvec} &= \diag\pare{\frac{1}{1+e^{-\zvec}}(1-\frac{1}{1+e^{-\zvec}})}\\
\Jac{\yvec}{\phi} &= \onevect
\end{split}
$$

And thus the Jacobian of the whole composition is given by:

$$
\begin{split}
\Jac{x}{}\phi(x) = \onevect\diag\pare{\frac{1}{1+e^{-\wvec x}}(1-\frac{1}{1+e^{-\wvec x}})}\wvec
\end{split}
$$

which yields a scalar, as expected since $m=n=1$ and thus the Jacobian is directly the derivative.
