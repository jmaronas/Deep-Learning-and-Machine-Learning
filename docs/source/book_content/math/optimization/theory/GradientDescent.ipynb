# Iterative Gradient methods: gradient descent and gradient ascent

Gradient descent is an optimization technique to find minima or maxima in a mathematical function through a numerical approximation. It is widely used within the machine learning community and is the basis for more advanced optimization techniques.

Gradient descent is referred to as finding the minimum, while gradient ascent is referred to as finding the maximum. In general, people often refer to these methods as iterative gradient-based methods, gradient-based methods, or simply gradient descent. 

Note that one often refers to these methods as gradient descent because running gradient ascent on a function is equivalent to running gradient descent over the same function multiplied by $-1$.

## Gradient descent on unidimensional functions

It turns out that some functions have maximums or minimums that cannot be obtained analytically. This implies that although, for example, the minimum exists and we can conclude it is a minimum, we cannot solve the resulting equations or system of equations.

For example consider the function $\phi : \mathbb{R} \to \mathbb{R}$ :

$$
\begin{split}
\phi(x) = x^2 + e^x
\end{split}
$$

```python
N_grid = 100
min_x = -5
max_x = 5
x_grid = np.linspace(min_x,max_x,N_grid)

plt.plot(x_grid, x_grid**2 + np.exp(x_grid))
```

The first derivative is given by:

$$
\begin{split}
\frac{\dd \phi}{\dd x} = 2x + e^x
\end{split}
$$

To find the minimum, we need to solve:

$$
2x + e^x = 0
$$

which does not have a solution in closed form (or at least I do not know how to obtain it). The second derivative:

$$
\begin{split}
2 + e^x > 0
\end{split}
$$

is always positive for any value of $x$, guaranteeing that the critical point is a minimum.

The existence of the minimum can be shown because when $x$ goes to minus or plus infinity, the function always increases and never decreases at any point.
    

In arbitrary dimensions  $\phi : \mathbb{R}^2 \to \mathbb{R}$  the function:

$$
\begin{split}
f(x) = x^2 + e^x + y^2
\end{split}
$$

has the same property.

```python
def fun1(x,y):
    return x**2 + np.exp(x) + y**2

## define grid
N_grid = 100
min_x = -2
max_x = 2
min_y = -2
max_y = 2

x_grid = np.linspace(min_x,max_x,N_grid)
y_grid = np.linspace(min_y, max_y, N_grid)

X_mesh, Y_mesh = np.meshgrid(x_grid, y_grid)

## define plot
fig = plt.figure(figsize=(12, 16))
ax_list = []

for i in range(1):
    ax1 = fig.add_subplot(4, 2, 2*i+1)
    ax2 = fig.add_subplot(4, 2, 2*i+2, projection='3d')
    
    ax_list.append([ax1,ax2])

# define set of func to show
funcs = [fun1]

for i in range(1):
    ax1,ax2 = ax_list[i]
    
    Z_mesh = funcs[i](X_mesh,Y_mesh)
    
    ax1.contourf(X_mesh,Y_mesh,Z_mesh, cmap = 'gray')
    ax2.plot_surface(X_mesh,Y_mesh,Z_mesh, cmap = 'gray')
    
```

Numerical optimization is a set of techniques designed to find minima and maxima when exact computation is not possible or computationally expensive.

The simplest and most widely adopted method for numerical optimization is what is called gradient descent. The idea is to find the minimum of a function by starting from a point and taking steps in the direction of the gradient.

Suppose we are at any point in our previous function.

```python
def fun(x):
    return x**2 + np.exp(x)

N_grid = 100

min_x = -5
max_x = 5
x_grid = np.linspace(min_x,max_x,N_grid)

x_init = 4

plt.plot(x_grid, fun(x_grid))
plt.plot(x_init, fun(x_init),'o', color = 'C1')
```

By definition, we know that the gradient or derivative at a point is the tangent function at that point, where the slope is the actual derivative. In other words:

```python
def fun(x):
    return x**2 + np.exp(x)

def grad_f(x):
    return 2*x + np.exp(x)

N_grid = 100

min_x = -5
max_x = 5
x_grid = np.linspace(min_x,max_x,N_grid)

min_grad = 2
max_grad = 5
grad_grid = np.linspace(min_grad,max_grad,N_grid)

x_init = 4
grad_init = grad_f(x_init)
fun_init = fun(x_init)

## get the gradient function at the point x_init (tangent at the point)
gradient_function_x_at_current_x = grad_init * grad_grid + fun_init - grad_init * x_init

plt.plot(x_grid, fun(x_grid), label = 'function')
plt.plot(x_init, fun(x_init),'o', color = 'C1')
plt.plot(grad_grid, gradient_function_x_at_current_x, color = "C2", label = f'gradient function at $x={x_init}$')
plt.text(2,-15, f'$\\frac{{\partial f(x)}}{{\partial x}} = {grad_init:.3f}$\n$y = {grad_init:.3f}x+{fun_init - grad_init * x_init:.3f}$', color = "C2")
plt.legend()
```

Since the direction of change of the curve goes up, the derivative is positive. In other words, the slope of the tangent line is positive. We can see that the minimum is exactly in the opposite direction to the sign of the derivative. This means that if we iteratively follow the negative direction of the gradient, we can reach the minimum. This suggests the following algorithm to find minima of functions:

* Take an initial guess for $x_t$.
* Repeat until the derivative is zero:

    * step 1: compute the gradient of the function at $x_t$
    * step 2: get a new guess $x_{t+1} = x_t - \alpha \cdot \frac{\partial f(x_t)}{\partial x_t}$
    
The term $\alpha$ is referred to as the learning rate and is the amount of the gradient considered in the update. Since gradient descent only uses the information of the gradient (which is the first-order derivative), it is also known as a first-order gradient-based method. In some cases, optimal values for $\alpha$ exist, which implies a single step can reach the minimum. There are also multiple variants of this algorithm as well.

Note that the value of $\alpha$ is necessary because, in the considered example at $x=4$ the gradient is $62.6$, so moving in the minus direction of the gradient by this amount would lead to a point that is far from the minimum $4-62.6 \approx -58$, while the minimum is around $0$.

Gradient ascent is exactly the same algorithm, but instead of subtracting the gradient, we add it. This follows the direction of the derivative and thus reaches a maximum

Let's visualize how gradient descent works.

```python
## ===================
## computational graph
def fun(x):
    return x**2 + np.exp(x)

def grad_f(x):
    return 2*x + np.exp(x)

## ===================
## plot configurations
N_grid = 100

min_x = -6
max_x = 6
x_grid = np.linspace(min_x,max_x,N_grid)

min_grad = -5
max_grad = 5
grad_grid = np.linspace(min_grad,max_grad,N_grid)

## ======
## figure
fig, (ax1, ax2) = plt.subplots(1,2, figsize = (15,5))

# Create temporary file for video creation
video_filename = "/tmp/aux.mp4"

## video writer
writer = imageio.get_writer(video_filename, format="FFMPEG", mode="I", fps=1, codec="libx264")

## ========
## initialization point and gradient descent parameters
x = 5
lr = 0.01
epochs = 10
history_epochs = []
history_fx = []

for e in range(epochs):
    
    ## compute gradient
    grad_x = grad_f(x)
    
    ## compute function 
    f_x = fun(x)

    ## get the gradient function at the point x_init (tangent at the point)
    gradient_function_x_at_current_x = grad_x * grad_grid + f_x - grad_x * x

    ## grad descent update
    x_n = x - lr * grad_x
    
    ## =============
    ## start display
    ax1.clear()
    
    ax1.set_xlim([-9,9])
    ax1.set_ylim([-100,200])
    
    ## 1.
    ## display loss function and current point
    ax1.plot(x_grid, fun(x_grid), label = 'function')
    ax1.plot(x, -50,'*', color = 'C1')
    ax1.text(x +0.5, -51 , f"x = {x:.2f}", fontsize=12, va='bottom', color = f"C1" , zorder = 50)
    ax1.legend(loc = "upper right")

    fig.canvas.draw()
    frame = np.asarray(fig.canvas.buffer_rgba())[:, :, :3]
    writer.append_data(frame)

    ## 2.
    ## display vertical line and current function value
    ax1.vlines(x, ymin=-50, ymax=f_x, color='k', linestyles='dotted', zorder = -50)

    ## 3. display current loss
    ax1.plot(x, f_x, 'o', color = 'C1', label = f'$f(x)$', zorder = 20)
    ax1.text(x + 0.5, f_x , f"$f(x)={f_x:.3f}$", fontsize=12, va='bottom', color = "C0" , zorder = 50)
    ax1.legend(loc = "upper right")
    
    fig.canvas.draw()
    frame = np.asarray(fig.canvas.buffer_rgba())[:, :, :3]
    writer.append_data(frame)
  
    ## 4 display tangent line
    ax1.plot(grad_grid, gradient_function_x_at_current_x, color = "C2", label = f'gradient function at $x={x_init}$')
    symbol = " "
    if f_x - grad_x * x > 0:
        symbol = "+"
    ax1.text(2,-15, f'$\\frac{{\partial f(x)}}{{\partial x}} = {grad_x:.3f}$\n$y = {grad_x:.3f}x{symbol}{f_x - grad_x * x:.3f}$', color = "C2")
    ax1.legend(loc = "upper right")
    
    fig.canvas.draw()
    frame = np.asarray(fig.canvas.buffer_rgba())[:, :, :3]
    writer.append_data(frame)

    ## 5.
    ## display rest of lines to show update
    ax1.hlines(y = f_x, xmin=x_n, xmax=x, color='k', linestyles='dotted', zorder = -50)
    ax1.vlines(x_n, ymin=-51 , ymax=f_x, color='k', linestyles='dotted', zorder = -50)
    
    fig.canvas.draw()
    frame = np.asarray(fig.canvas.buffer_rgba())[:, :, :3]
    writer.append_data(frame)

    ## 6. display new point
    ax1.plot(x_n, -55, '*', color = 'C3', label = r'updated value: $x_n = x - \alpha\cdot\frac{\partial f(x)}{\partial x}$', zorder = 200, markersize = 10)
    ax1.text(x_n+0.5, -63, f"$x_n = {x:.2f} -{lr:.2f}*{grad_x:.2f} = {x_n:.2f}$", va='bottom', color = f"C3" , zorder = 200) 
    ax1.legend(loc = "upper right")
    
    fig.canvas.draw()
    frame = np.asarray(fig.canvas.buffer_rgba())[:, :, :3]
    writer.append_data(frame)

    ax1.set_title(f"$f(x)={f_x:.3f}$ \n $f(x_t) = {fun(x_n):.3f}$")
      
    fig.canvas.draw()
    frame = np.asarray(fig.canvas.buffer_rgba())[:, :, :3]
    writer.append_data(frame)

    ## =============
    ## Plot function minimization over the course of learning
    ax2.clear()
    history_epochs.append(e)
    history_fx.append(f_x)
    
    ax2.plot(history_epochs, history_fx,'o', color = 'k')
    ax2.plot(history_epochs, history_fx, color = 'k')
    
    ax2.set_xlabel('Epochs')
    ax2.set_ylabel('$f(x)$')
    ax2.set_xlim([-1,epochs+1])
    ax2.set_ylim([0,200])

    fig.canvas.draw()
    frame = np.asarray(fig.canvas.buffer_rgba())[:, :, :3]
    writer.append_data(frame)

    # to repeat loop
    x = x_n
    
writer.close() 
plt.close()
```

### The role of the learning rate

Very big learning rates will make gradient descent not converge to the minimum. Very small learning rates will make convergence slow. The optimal learning rate depends on the curvature of the function. There are more advanced gradient descent methods which use second-order information to adjust the learning rate depending on the curvature. When the curvature is smaller, the learning rate may increase. These are known as second-order methods. We will see that there are more sophisticated gradient descent methods that make use of some gradient statistics to improve convergence.

### Optimal learning rate depends on the curvature of the function.

Higher curvature requires a smaller learning rate, while lower curvature requires a higher one. This implies that we can only precompute an optimal learning rate for functions with constant curvature (quadratic ones).

In one dimension, the optimal learning rate is given by the inverse of the curvature. Thus, at each point, we can compute the second derivative and set the learning rate to 1 over the second derivative at that point.

```python
## ===================
## computational graph
def fun(x):
    return x**4 

def grad_f(x):
    return 4*x**3

def grad2_f(x):
    return 12*x**2

## ===================
## plot configurations
N_grid = 100

min_x = -4
max_x = 4
x_grid = np.linspace(min_x,max_x,N_grid)

min_grad = -4
max_grad = 4
grad_grid = np.linspace(min_grad,max_grad,N_grid)

## ========
## initialization point and grad descent parameters
xs = [3,3]
lrs = [0.009,0.009]
epochs = 6
adaptative_lrs = [True,False]

## ======
## figure
fig, axs = plt.subplots(1,len(adaptative_lrs), figsize = (10,5))
if len(lrs) == 1:
    axs = [axs]

# Create temporary file for video creation
video_filename = "/tmp/aux.mp4"

## video writer
writer = imageio.get_writer(video_filename, format="FFMPEG", mode="I", fps=1, codec="libx264")

for e in range(epochs):
    
    ## get optimal lr
    for adaptative_lr in adaptative_lrs:
        if adaptative_lr:
            lrs[0] = 1/grad2_f(xs[0])
    
    ## compute gradient
    grad_xs = []
    for x in xs:
        grad_xs.append( grad_f(x) )
    
    ## compute function 
    f_xs = []
    for x in xs:
        f_xs.append(fun(x))

    ## get the gradient function at the point x_init (tangent at the point)
    gradient_function_x_at_current_xs = []
    for f_x,x,grad_x in zip(f_xs,xs, grad_xs):
        gradient_function_x_at_current_xs.append(grad_x * grad_grid + f_x - grad_x * x)

    x_ns = []
    for lr,x,grad_x in zip(lrs,xs,grad_xs):
        ## grad descent update
        x_ns.append(x - lr * grad_x)
    
    ## =============
    ## start display
    for ax in axs:
        ax.clear()
    
        ax.set_xlim([-5,5])
        ax.set_ylim([-20,80])
        
    for lr,ax in zip(lrs,axs):
        ax.set_title(f"lr={lr:.3f}")
    
    ## 1.
    ## display loss function and current point
    for x,ax in zip(xs,axs):
        ax.plot(x_grid, fun(x_grid), label = 'function')
        ax.plot(x, -5,'*', color = 'C1')
        ax.text(x +0.5, -6 , f"x = {x:.2f}", fontsize=12, va='bottom', color = f"C1" , zorder = 50)
        ax.legend(loc = "upper right")

    fig.canvas.draw()
    frame = np.asarray(fig.canvas.buffer_rgba())[:, :, :3]
    writer.append_data(frame)

    ## 2.
    ## display vertical line and current function value
    for x,f_x,ax in zip(xs,f_xs,axs):
        ax.vlines(x, ymin=-6, ymax=f_x, color='k', linestyles='dotted', zorder = -50)
 
        ## 3. display current loss
        ax.plot(x, f_x, 'o', color = 'C1', label = f'$f(x)$', zorder = 20)
        ax.text(x + 0.5, f_x , f"$f(x)={f_x:.3f}$", fontsize=12, va='bottom', color = "C0" , zorder = 50)
        ax.legend(loc = "upper right")
    

    fig.canvas.draw()
    frame = np.asarray(fig.canvas.buffer_rgba())[:, :, :3]
    writer.append_data(frame)
  
    
    ## 4 display tangent line
    for ax,gradient_function_x_at_current_x,grad_x,f_x,x in zip(axs,gradient_function_x_at_current_xs,grad_xs,f_xs,xs):
        ax.plot(grad_grid, gradient_function_x_at_current_x, color = "C2", label = f'gradient function at $x={x}$')
        symbol = " "
        if f_x - grad_x * x > 0:
            symbol = "+"
        ax.text(2,-8, f'$\\frac{{\partial f(x)}}{{\partial x}} = {grad_x:.3f}$\n$y = {grad_x:.3f}x{symbol}{f_x - grad_x * x:.3f}$', color = "C2")
        ax.legend(loc = "upper right")
    
    fig.canvas.draw()
    frame = np.asarray(fig.canvas.buffer_rgba())[:, :, :3]
    writer.append_data(frame)

    
    ## 5.
    ## display rest of lines to show update
    for x,x_n,f_x,ax in zip(xs,x_ns,f_xs,axs):
        ax.hlines(y = f_x, xmin=x_n, xmax=x, color='k', linestyles='dotted', zorder = -50)
        ax.vlines(x_n, ymin=-10, ymax=f_x, color='k', linestyles='dotted', zorder = -50)
    

    fig.canvas.draw()
    frame = np.asarray(fig.canvas.buffer_rgba())[:, :, :3]
    writer.append_data(frame)

    
    ## 6. display new point
    for x,x_n,grad_x,lr,ax in zip(xs,x_ns,grad_xs,lrs,axs):
        ax.plot(x_n, -12, '*', color = 'C3', label = r'updated value: $x_n = x - \alpha\cdot\frac{\partial f(x)}{\partial x}$', zorder = 200, markersize = 10)
        ax.text(x_n+0.5, -14, f"$x_n = {x:.2f} -{lr:.2f}*{grad_x:.2f} = {x_n:.2f}$", va='bottom', color = f"C3" , zorder = 200) 
        ax.legend(loc = "upper right")
    

    fig.canvas.draw()
    frame = np.asarray(fig.canvas.buffer_rgba())[:, :, :3]
    writer.append_data(frame)

    
    for f_x,x_n,ax in zip(f_xs,x_ns,axs):
        ax.set_title(f"$f(x)={f_x:.3f}$ \n $f(x_n) = {fun(x_n):.3f}$")
      
    fig.canvas.draw()
    frame = np.asarray(fig.canvas.buffer_rgba())[:, :, :3]
    writer.append_data(frame)
    
    # to repeat loop
    xs = x_ns
    
    
writer.close() 
plt.close()
```

## Gradient descent on multidimensional functions

Gradient descent can obviously run on multidimensional functions by replacing the derivative with the Jacobian (or gradient, since it is its transpose). Current libraries use the Jacobian since automatic differentiation is based on vector Jacobian computations.

Here the role of the second derivative is played by the Hessian. We can use it to get optimal learning rates in each direction or, equivalently, to precondition the gradient by the inverse Hessian to get optimal updates. These are known as second-order methods and are computationally expensive for many models of interest.

```python
def fun(x,y):
    return x**4 -2*x**2 + 2*y**2

def grad_fun(x,y):
    return 4*x**3 - 4*x, 4*y


## define grid
N_grid = 100
min_x = -2
max_x = 2
min_y = -2
max_y = 2

x_grid = np.linspace(min_x,max_x,N_grid)
y_grid = np.linspace(min_y, max_y,N_grid)

X_mesh, Y_mesh = np.meshgrid(x_grid, y_grid)
Z_mesh = fun(X_mesh,Y_mesh)

## define vector field
step = 5
Xq = X_mesh[::step, ::step]
Yq = Y_mesh[::step, ::step]
Uq, Vq = grad_fun(Xq, Yq)

## define plot
fig = plt.figure(figsize=(10, 5))
ax1 = fig.add_subplot(1, 2, 1)
ax2 = fig.add_subplot(1, 2, 2, projection='3d')
ax2.view_init(elev=65)

# Initialize parameters
x = -1.7
y = -1.7

## gradient descent parameters
lr = 0.1
epochs = 10


# Create temporary file for video creation
video_filename = "/tmp/aux.mp4"

## video writer
writer = imageio.get_writer(video_filename, format="FFMPEG", mode="I", fps=1, codec="libx264")


for e in range(epochs):

    ## forward plus backward
    grad_x, grad_y = grad_fun(x,y)

    ## compute function at current parameter value
    f_xy = fun(x,y)

    ## compute function on updated parameters
    x_n = x-lr*grad_x
    y_n = y-lr*grad_y

    ## function on new parameters
    f_xy_n = fun(x_n,y_n)

    ## ============= ##
    ## START DRAWING ##
    ## ============= ##
    ax1.clear()
    ax2.clear()

    ## 1 display function
    ax1.contourf(X_mesh,Y_mesh,Z_mesh, cmap = 'gray')
    ax2.plot_surface(X_mesh,Y_mesh,Z_mesh, alpha=0.8, cmap = 'gray')

    ## display vector field
    ax1.quiver(Xq, Yq, Uq, Vq, color='green')
    
    ## 2 display current point
    ax1.plot(x,y, '*', color='C1')
    ax1.text(x, y, f"(x,y) = ({x:.2f},{y:.2f})", fontsize=12, va='bottom', color="C1")

    ax2.scatter(x,y,f_xy, '*', color='C1')
    ax2.text(x, y, f_xy, f"(x,y) = ({x:.2f},{y:.2f})", fontsize=12, va='bottom', color="C1")

    ## 3 display the gradient arrow and updated weight
    a = Arrow3D([x, x_n], [y, y_n], [f_xy, f_xy_n],
                mutation_scale=20, lw=0.5, arrowstyle="-|>", color="C0")
    ax2.add_artist(a)
    
    ## 4 display new weight
    ax1.plot(x_n, y_n, 'o', color='C3')
    ax1.text(x_n, y_n, f"$(x_n,y_n) = ({x_n:.2f},{y_n:.2f})$", fontsize=12, va='bottom', color="C3")

    ax1.arrow(x, y , x_n-x, y_n-y,
              color="red",
              width=0.02,
              head_width=0.1,
              head_length=0.03,
              length_includes_head=True)

    ax2.scatter(x_n, y_n, f_xy_n, 'o', color='C3')
    ax2.text(x_n, y_n, f_xy_n, f"$(x_n,y_n) = ({x_n:.2f},{y_n:.2f})$", fontsize=12, va='bottom', color="C3")

    ## save frame
    fig.canvas.draw()
    frame = np.asarray(fig.canvas.buffer_rgba())[:, :, :3]
    writer.append_data(frame)

    x = x_n
    y = y_n

writer.close()
plt.close()
    
```
